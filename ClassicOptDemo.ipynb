```python
import numpy as np
import matplotlib.pyplot as plt


x = [0, .5, 1, 1.5, 2, 2.5, 3, 3.5, 4]

def f(x):
  return 4-(x-2)**2

x_vals = np.linspace(0,4,100)
y_vals = f(x_vals)
plt.plot(x_vals, y_vals)
plt.show()
```

```python

#simple guess and check loop

max_h = 0


#loop through
for x_guess in x:
  height = f(x_guess)
  print(height)
  if height > max_h:
    max_h = height
result = (f"the max height is {max_h} ")
print(result)

#find max height
```

```python
#Gradient ascent

#define function
def function():
  f = 4 - (x-2)**2
  return f

#define derivative (If using a more complex function a calculator may be useful)
def df(x):
  return -2 * (x-2)

#parameters
x = 0 # Initial guess for the max value of x (for starting purpose)
step = 0.1 # Learning rate or step size for each iteration (note: making this negetive will only search left)
epochs = 50 # Number of iterations to run the gradient ascent algorithm

for epoch in range(epochs):
  grad = df(x)
  x = x + step * grad
  print(f"Step {epoch}: x = {x:.4f}")
```
